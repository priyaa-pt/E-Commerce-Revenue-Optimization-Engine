### Load engineered data

```python
import pandas as pd

df = pd.read_csv("engineered_data.csv")
df['order_date'] = pd.to_datetime(df['order_date'])
```

### Sort by time

```python
df = df.sort_values('order_date')

```

### Apply time-based split

```python
split_date = "2024-09-01"

train = df[df['order_date'] < split_date]
test  = df[df['order_date'] >= split_date]

print(train.shape, test.shape)
```

### Define target & features

```python
target = 'daily_demand'
features = [col for col in df.columns if col not in ['daily_demand', 'order_date']]
```

### Save splits


```python
train.to_csv("train_data.csv", index=False)
test.to_csv("test_data.csv", index=False)
```
